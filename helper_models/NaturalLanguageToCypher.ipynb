```python
import pandas as pd
import numpy as numpy


```

```python
df = pd.DataFrame(pd.read_csv("data/cypher_dataset.csv"))
df_corpus=''.join(df['Human Prompt']) + ''.join(df['Cypher Query'])
#cannot strip special characters as they form syntax of the cypher queries
```

```python
#df_vocab = set(df_corpus.split(" "))  this splits into words
df_vocab = set(df_corpus)
#df_vocab
```

```python
char_to_int = {char : i for i, char in enumerate(df_vocab)}
char_to_int

int_to_char = {i : char for i, char in enumerate(df_vocab)}
#int_to_char
```

```python
df['prompt_encoded'] = df['Human Prompt'].apply(lambda entry: [char_to_int[character] for character in entry])
#df['prompt_encoded'].head()

df['cypher_encoded'] = df['Cypher Query'].apply(lambda entry : [char_to_int[character] for character in entry])
```
